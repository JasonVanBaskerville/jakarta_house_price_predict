# Jakarta House Price Predict

Dataset ini memuat data harga rumah yang dikumpulkan dari daftar properti di Jakarta, Indonesia.

Data tersebut diperoleh melalui teknik *web scraping* dari Rumah123.com dan mencakup 10.000 data properti. Setiap data memuat informasi seperti lokasi, harga rumah, jumlah ruangan, luas tanah, dan luas bangunan.

Web scraping = teknik otomatis untuk mengekstrak atau mengambil data dari sebuah website dalam jumlah besar secara cepat.

1. index = Pengenal unik untuk setiap catatan data
2. price = Harga rumah
3. district = Kecamatan tempat rumah berada
4. city = Kota tempat rumah berada
5. bed_rooms = Jumlah kamar tidur
6. bath_rooms = Jumlah kamar mandi
7. carport = Jumlah carport (garasi mobil)
8. land_area = Luas tanah rumah (dalam meter persegi)
9. building_area = Luas bangunan rumah (dalam meter persegi)

Dataset ini sangat tepat digunakan untuk analisis pasar real estat dan pemodelan prediktif, karena memberikan wawasan mengenai faktor-faktor yang memengaruhi harga properti di Jakarta, Indonesia.

## Menentukan Target dan Features

```python
X = data[['district', 'city', 'bed_rooms', 'bath_rooms', 'carport', 'building_area', 'land_area']]
y = data[["price"]]
```

```python
X
```

## Memisahkan Train dan Test

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
)
```

## Preprocessing

```python
numeric = [
    "bed_rooms",
    "bath_rooms",
    "carport",
    "land_area",
    "building_area"
]
```

```python
categorical = [
    "district",
    "city"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(
            drop="first",
            handle_unknown="ignore"
        ), categorical)
    ],
    remainder="drop"
)
```

## Buat Model

### Model 1. Linear Regression

#### Buat model

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

# Buat pipeline
pipeline_lr = Pipeline([
    ("prep", preprocessor),
    ("lr", lr)
])
```

#### Training

```python
pipeline_lr.fit(X_train, y_train)
```

#### Prediksi

```python
y_pred_lr = pipeline_lr.predict(X_test)
y_pred_lr
```

#### Evaluasi model

```python
# if R = 1 -> sangat baik
from sklearn.metrics import r2_score

r_squared_lr = r2_score(y_test, y_pred_lr)

print(f'R-squared: {r_squared_lr}')
```

```python
# if MAE = 0 -> sangat baik
# if MSE = 0 -> sangat baik
# Gunakan MSE ketika: kesalahan besar harus diberikan penalti lebih berat

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error

MAE_lr = mean_absolute_error(y_test, y_pred_lr)
MSE_lr = mean_squared_error(y_test, y_pred_lr)

print(f'MAE: {MAE_lr}')
print(f'MSE: {MSE_lr}')
```

### Model 2. Decision Tree Regressor

#### Buat model

```python
from sklearn.model_selection import GridSearchCV

# GridSearchCV digunakan untuk mencari kombinasi hyperparameter terbaik dari sebuah model Machine Learning secara otomatis.
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor()
tree_model
```

```python
pipeline_tree = Pipeline([
    ("prep", preprocessor),
    ("model", tree_model)
])
```

```python
param_grid = {
    "model__criterion": [
        "squared_error",
        "absolute_error",
        "poisson"
    ],
    
    "model__max_depth": [
        None,
        5,
        10,
        15,
        20
    ],
    
    "model__min_samples_split": [
        2,
        5,
        10
    ],
    
    "model__min_samples_leaf": [
        1,
        2,
        5
    ]
}
```

```python
grid_tree = GridSearchCV(
    estimator=pipeline_tree,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)

grid_tree
```

#### Training

```python
grid_tree.fit(X_train, y_train)
```

```python
grid_tree.best_params_
```

#### Prediksi

```python
# y prediksi
tree_preds = grid_tree.predict(X_test)
tree_preds
```

#### Evaluasi Model

```python
# if R = 1 -> sangat baik
r_squared_tree = r2_score(y_test, tree_preds)

print(f'R-squared: {r_squared_tree}')
```

```python
# if MAE = 0 -> sangat baik
# if MSE = 0 -> sangat baik
# Gunakan MSE ketika: kesalahan besar harus diberikan penalti lebih berat

MAE_tree = mean_absolute_error(y_test, tree_preds)
MSE_tree = mean_squared_error(y_test, tree_preds)

print(f'MAE: {MAE_tree}')
print(f'MSE: {MSE_tree}')
```

### Model 3. Random Forest Regressor

#### Buat model

```python
from sklearn.ensemble import RandomForestRegressor
rfr_model = RandomForestRegressor()
```

```python
pipeline_rfr = Pipeline([
    ("prep", preprocessor),
    ("model", rfr_model)
])
```

```python
param_grid_rfr = {
    "model__n_estimators": [2, 3, 4, 5, 6, 7, 8, 9, 10],
    "model__max_depth": [None, 5, 10, 15]
}
```

```python
grid_rfr = GridSearchCV(
    estimator=pipeline_rfr,
    param_grid=param_grid_rfr,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)

grid_rfr
```

#### Training

```python
grid_rfr.fit(X_train, y_train)
```

```python
grid_rfr.best_params_
```

#### Prediksi

```python
rfr_preds = grid_rfr.predict(X_test)
```

#### Evaluasi Model

```python
# if R = 1 -> sangat baik
r_squared_rfr = r2_score(y_test, rfr_preds)

print(f'R-squared: {r_squared_rfr}')
```

```python
# if MAE = 0 -> sangat baik
# if MSE = 0 -> sangat baik
# Gunakan MSE ketika: kesalahan besar harus diberikan penalti lebih berat

MAE_rfr = mean_absolute_error(y_test, rfr_preds)
MSE_rfr = mean_squared_error(y_test, rfr_preds)

print(f'MAE: {MAE_rfr}')
print(f'MSE: {MSE_rfr}')
```

## Bandingkan hasil evaluasi model

```python
model_evaluation = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree Regressor",
        "Random Forest Regressor"
    ],
    "R²": [
        r_squared_lr,
        r_squared_tree,
        r_squared_rfr
    ],
    "MAE": [
        MAE_lr,
        MAE_tree,
        MAE_rfr
    ],
    "MSE": [
        MSE_lr,
        MSE_tree,
        MSE_rfr
    ]
})

model_evaluation
```

```python
model_evaluation.style.format({
    "R²": "{:.4f}",
    "MAE": "{:,.2f}",
    "MSE": "{:,.2f}"
}).highlight_min(
    subset=["MAE", "MSE"],
    axis=0
).highlight_max(
    subset=["R²"],
    axis=0
)
```
